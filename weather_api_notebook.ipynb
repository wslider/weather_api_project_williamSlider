```python
def get_weather(city):
    url = f"http://api.weatherapi.com/v1/current.json?key={api_key}&q={city}"
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        current = data['current']
        location = data['location']
        temp_f = current['temp_f']
        wind_mph = current['wind_mph']
        humidity = current['humidity']
        print(f"Weather for {location['name']}, {location['region']}:")
        print(f"Temperature : {temp_f}°F")
        print(f"Humidity    : {humidity}%")
        print(f"Wind        : {wind_mph} mph")
    else: 
        print(f"error fretching data for {city}. {response.status_code}!")


    
    
```

```python
cities = ["Louisville", "Trivandrum", "Reykjavik", "Dubai"]
weather_data = []

for city in cities:
    url = f"http://api.weatherapi.com/v1/current.json?key={api_key}&q={city}"
    response = requests.get(url)
    
    if response.status_code == 200:
        data = response.json()
        weather_data.append({
            "city": data['location']['name'],
            "condition": data['current']['condition']['text'],
            "temperature" : data['current']['temp_f'], 
            "humidity": data['current']['humidity'],
            "wind speed": data['current']['wind_mph'],
        })
    else: 
        print(f"error fetching data for {city}. {response.status_code}")

weather_data
    
```

```python
api_key = os.getenv("API_KEY")

def get_city_data(city, api_key):
    baseurl = "http://api.weatherapi.com/v1/"
    today = pd.Timestamp.now().strftime("%Y-%m-%d")
    current_url = f"{baseurl}current.json?key={api_key}&q={city}"
    forecast_url = f"{baseurl}forecast.json?key={api_key}&q={city}&days=3"
    astronomy_url = f"{baseurl}astronomy.json?key={api_key}&q={city}&dt={today}"

    # get the current weather

    curr_wx_response = requests.get(current_url)
    curr_wx_data = curr_wx_response.json()
    curr_condition = curr_wx_data['current']['condition']['text']
    curr_temp_f = curr_wx_data['current']['temp_f']  
    curr_humitity = curr_wx_data['current']['humidity']
    curr_wind_mph = curr_wx_data['current']['wind_mph']
    curr_wx_summary = f"{curr_condition} - Temp: {curr_temp_f}°F  Humidity: {curr_humitity}%  Wind: {curr_wind_mph}MPH"
    

    # get a forecast - loop through to get all three days 

    forecast_response = requests.get(forecast_url)
    forecast_data = forecast_response.json()
    forecast = forecast_data['forecast']['forecastday']

    forecast_3_days_summary = []
    for day in forecast:
        date_obj = datetime.strptime(day['date'], '%Y-%m-%d')
        formated_date = date_obj.strftime('%A, %B, %#d')
        forecast_3_days_summary.append({
        f"{formated_date}: High {day['day']['maxtemp_f']}F / Low {day['day']['mintemp_f']}F - {day['day']['condition']['text']}"
        })

    # get sunrise / sunset
    astro_response = requests.get(astronomy_url)
    astronomy_data = astro_response.json()
    sunrise = astronomy_data['astronomy']['astro']['sunrise']
    sunset = astronomy_data['astronomy']['astro']['sunset']
    moonphase = astronomy_data['astronomy']['astro']['moon_phase']

    astro_summary = f"Sunrise: {sunrise}  Sunset: {sunset}  Moon: {moonphase}"

    # build a summary
    city_full_wx_summary = f"Currently: {curr_wx_summary}, Forecast: {forecast_3_days_summary}, Astronomy: {astro_summary}"
    return city_full_wx_summary


cities = ["Louisville", "Trivandrum", "Reykjavik", "Dubai"]
weather_summaries = []

for city in cities:
    city_full_wx_summary = get_city_data(city, api_key)
    weather_summaries.append({
        'city': city, 
        'summary': city_full_wx_summary, 
    })

weather_summaries
    

# save into a dictionary -> df -> csv file 
weather_summaries_df = pd.DataFrame(weather_summaries)
weather_summaries_df.to_csv(f"{today}_cities_wx_data.csv", index=False)
```
